## Section A: Understanding the Dataset

### 1. Source and Purpose
* **Source:** The dataset originates from the **2015 Behavioral Risk Factor Surveillance System (BRFSS)**, an annual telephone survey conducted by the US **Centers for Disease Control and Prevention (CDC)**.
* **Purpose:** It is used to track health risk behaviors, clinical conditions, and preventive services in the population. For this assignment, we use it to study how physiological indicators (BMI, blood pressure, cholesterol) and behavioral habits (physical activity) relate to diabetes status.

### 2. Meaning and Coding of Selected Variables
* **`Diabetes_012`**: Target disease indicator.
  * `0` = No diabetes
  * `1` = Prediabetes
  * `2` = Diabetes
* **`BMI`**: Body Mass Index (continuous numerical indicator calculated as $\text{Weight (kg)} / \text{Height (m)}^2$).
* **`HighBP`**: High Blood Pressure status.
  * `0` = No high blood pressure
  * `1` = High blood pressure
* **`HighChol`**: High Blood Cholesterol status.
  * `0` = No high cholesterol
  * `1` = High cholesterol
* **`PhysActivity`**: Physical activity in the past 30 days.
  * `0` = No physical activity
  * `1` = Active
* **`Age`**: Imputed age category (1 to 13, representing 5-year intervals starting from 18-24 up to 80+).

### 3. Limitations of the Dataset
* **Self-Reported Bias:** Since the survey is telephone-based, respondents may inaccurately report weight, height, or physical activity.
* **Cross-Sectional Limitation:** Because data is collected at a single point in time, we can only establish **statistical association**, not **causation**.
* **Geographical and Temporal Constraint:** US data from 2015 may not fully reflect the current epidemiological reality of diabetes risk profiles in Zambia.

## Section B: Dataset Preparation

### 1. Clinical Importance of Selected Variables
* **BMI (Body Mass Index):** Elevated BMI indicates excess adipose tissue, which is directly linked to insulin resistance—the hallmark of Type 2 Diabetes.
* **HighBP (Hypertension) & HighChol (Hypercholesterolemia):** Together with diabetes, these form "Metabolic Syndrome," sharing underlying pathogenetic mechanisms of endothelial dysfunction and cardiovascular risks.
* **PhysActivity:** Exercise increases skeletal muscle glucose uptake independent of insulin, serving as a critical protective factor.
* **Age:** Beta-cell function and insulin sensitivity naturally decline with age, making age an important non-modifiable risk factor.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats
import os
import zipfile

# Configure visual settings for university presentation slides
sns.set_theme(style="whitegrid")
plt.rcParams.update({'font.size': 12, 'axes.labelsize': 14, 'axes.titlesize': 16, 'xtick.labelsize': 11, 'ytick.labelsize': 11})

zip_path = '/content/diabetes_012_health_indicators_BRFSS2015 2.csv.zip'
extract_dir = '/content/extracted_data'
file_name = 'diabetes_small.csv'

# 1. Unzip the file if it exists and we haven't created the small CSV yet
if os.path.exists(zip_path) and not os.path.exists(file_name):
    print("📦 Unzipping dataset...")
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(extract_dir)

    # Locate the extracted CSV
    extracted_files = os.listdir(extract_dir)
    csv_files = [f for f in extracted_files if f.endswith('.csv')]

    if csv_files:
        large_csv_path = os.path.join(extract_dir, csv_files[0])
        print(f"Reading large dataset: {large_csv_path}")
        df_large = pd.read_csv(large_csv_path)

        # Create a representative small sample of 5,000 rows with a fixed random seed
        df_small = df_large.sample(n=5000, random_state=42).reset_index(drop=True)
        df_small.to_csv(file_name, index=False)
        print(f"✅ Generated '{file_name}' with {len(df_small)} rows.\n")
    else:
        print("❌ No CSV file found inside the zip.")

# 2. Load and verify our active dataset
if os.path.exists(file_name):
    df = pd.read_csv(file_name)
    print(f"✅ Successfully loaded '{file_name}' with {len(df)} rows and {df.shape[1]} columns.\n")

    # Select variables specified in Section B
    target_cols = ['BMI', 'Diabetes_012', 'HighBP', 'HighChol', 'PhysActivity', 'Age']
    available_cols = [col for col in target_cols if col in df.columns]

    print(f"Selected columns: {available_cols}")
    df = df[available_cols].copy()

    # Check for missing values
    missing_values = df.isnull().sum()
    print("\n--- Missing Values Count ---")
    print(missing_values)

    # Display basic descriptive overview
    print("\n--- Prepared Data Overview ---")
    display(df.head())
else:
    print(f"❌ ERROR: '{file_name}' could not be created or found.")
```
